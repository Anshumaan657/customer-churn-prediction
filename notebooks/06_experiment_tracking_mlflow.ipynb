### Phase 6 focuses on experiment tracking using MLflow.

MLflow was introduced to systematically track machine learning experiments, including model parameters, evaluation metrics, model artifacts, and run metadata. This helps compare multiple models in a reproducible and organized way.

In this project, MLflow is used after preprocessing, model training, and hyperparameter tuning. The processed datasets and saved model artifacts are reused to avoid repeating preprocessing unnecessarily.

The main goal of MLflow in this phase is to compare baseline and tuned models, especially Tuned Random Forest and Tuned XGBoost, using business-relevant metrics such as Recall, F1 Score, and ROC-AUC.

```python
import pandas as pd
import numpy as np
import joblib
import mlflow
import mlflow.sklearn
import mlflow.xgboost

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
```

```python
from pathlib import Path
import os
import mlflow

# Always resolve paths from the project root, even when this notebook is run from notebooks/.
PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

MLFLOW_DB_PATH = PROJECT_ROOT / "mlflow.db"
MLFLOW_ARTIFACT_ROOT = PROJECT_ROOT / "mlruns"
MLFLOW_TRACKING_URI = f"sqlite:///{MLFLOW_DB_PATH}"

os.environ["MLFLOW_TRACKING_URI"] = MLFLOW_TRACKING_URI
mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)

EXPERIMENT_NAME = "Customer Churn Prediction"
if mlflow.get_experiment_by_name(EXPERIMENT_NAME) is None:
    mlflow.create_experiment(
        EXPERIMENT_NAME,
        artifact_location=str(MLFLOW_ARTIFACT_ROOT)
    )

mlflow.set_experiment(EXPERIMENT_NAME)

print("MLflow tracking URI:", mlflow.get_tracking_uri())
print("MLflow experiment:", EXPERIMENT_NAME)
print("Artifact root:", MLFLOW_ARTIFACT_ROOT)
```

```python
with mlflow.start_run(run_name="mlflow_setup_test"):
    mlflow.log_param("test_param", "setup_check")
    mlflow.log_metric("test_metric", 1.0)
```
