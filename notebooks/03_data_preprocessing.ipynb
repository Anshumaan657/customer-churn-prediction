## 3.1 Missing Value Treatment – Hidden Missing Category Analysis

Although the dataset contained **no true missing values (`NaN`)**, an additional investigation was performed to identify **hidden missing categories** within categorical variables.

The following categorical features were examined:

- Gender
- Education_Level
- Marital_Status
- Income_Category
- Card_Category

### Findings

#### Education_Level

The `Education_Level` feature contained an `"Unknown"` category.

| Category | Count |
|-----------|--------|
| Graduate | 3128 |
| High School | 2013 |
| Unknown | 1519 |
| Uneducated | 1487 |
| College | 1013 |
| Post-Graduate | 516 |
| Doctorate | 451 |

---

#### Marital_Status

The `Marital_Status` feature also contained an `"Unknown"` category.

| Category | Count |
|-----------|--------|
| Married | 4687 |
| Single | 3943 |
| Unknown | 749 |
| Divorced | 748 |

---

#### Income_Category

The `Income_Category` feature contained an `"Unknown"` category as well.

| Category | Count |
|-----------|--------|
| Less than $40K | 3561 |
| $40K - $60K | 1790 |
| $80K - $120K | 1535 |
| $60K - $80K | 1402 |
| Unknown | 1112 |
| $120K + | 727 |

---

### Interpretation

The presence of `"Unknown"` values does not represent true missing data from a technical perspective. Instead, these categories indicate that the corresponding customer information was either unavailable or undisclosed.

In banking and customer analytics applications, such categories may themselves carry predictive information. For example, customers who choose not to disclose their income or educational background may exhibit behavioral patterns different from those who provide complete information.

Therefore, treating these observations as ordinary missing values and imputing or removing them could lead to the loss of potentially valuable signals.

------

### Decision

The `"Unknown"` categories will be **retained as valid categories** during preprocessing.

No imputation or row removal will be performed for these observations.

### Final Outcome of Missing Value Treatment

| Component | Decision |
|------------|-----------|
| True Missing Values (`NaN`) | No treatment required |
| Numerical Imputation | Not required |
| Categorical Imputation | Not required |
| Row Deletion | Not performed |
| `"Unknown"` Categories | Retained as valid categories |

### Business Justification

Preserving `"Unknown"` categories enables the model to learn whether the absence of certain customer information is associated with customer churn. This approach aligns with industry practices commonly adopted in financial and customer retention modeling.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
numerical_features = [
    "Customer_Age",
    "Dependent_count",
    "Months_on_book",
    "Total_Relationship_Count",
    "Months_Inactive_12_mon",
    "Contacts_Count_12_mon",
    "Credit_Limit",
    "Total_Revolving_Bal",
    "Avg_Open_To_Buy",
    "Total_Amt_Chng_Q4_Q1",
    "Total_Trans_Amt",
    "Total_Trans_Ct",
    "Total_Ct_Chng_Q4_Q1",
    "Avg_Utilization_Ratio"
]

categorical_features = [
    "Gender",
    "Education_Level",
    "Marital_Status",
    "Income_Category",
    "Card_Category"
]
```

```python
numeric_transformer = Pipeline(steps=[
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

### Now, let's just fit the preprocessor only on training data.

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

### fit_transform(X_train):
 learns scaling + category mapping from training data only

### transform(X_test):
 applies the same learned rules to unseen test data

```python
print("X_train_processed shape:", X_train_processed.shape)
print("X_test_processed shape:", X_test_processed.shape)
```

```python
# Get final feature names after preprocessing

feature_names = preprocessor.get_feature_names_out()

feature_names
```

```python
X_train_processed_df = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_processed_df = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)

X_train_processed_df.head()
```
